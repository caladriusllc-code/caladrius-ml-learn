```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
data = pd.read_csv("..\\src\\data\\DAX - Données Historiques.csv")
data.head(10)
```

```python
data['Variation %']
```

```python
sns.set_theme(style="whitegrid")

data["Rendement"] = pd.to_numeric(
    data["Variation %"].str.replace(",", ".", regex=False).str.rstrip("%")
) / 100
data["Date"] = pd.to_datetime(data["Date"], dayfirst=True)

cmap = sns.cubehelix_palette(rot=-.2, as_cmap=True)
g = sns.relplot(
    data=data,
    x="Date", y="Rendement", hue="Rendement", size="Rendement",
    palette=cmap, sizes=(10, 200),
)
g.ax.xaxis.grid(True, "minor", linewidth=.25)
g.ax.yaxis.grid(True, "minor", linewidth=.25)
g.despine(left=True, bottom=True)
```
