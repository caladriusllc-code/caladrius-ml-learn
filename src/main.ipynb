```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
data = pd.read_csv('./data_prepared.csv')

data.tail(20)
```

```python
#Work on signal about trend
data['MA50Signal'] = np.where(data['Dernier'] > data['MA50'], 1, 0)
data['MA200Signal'] = np.where(data['Dernier'] > data['MA200'], 1, 0)

conditions = [
    (data['Dernier'] > data['MA50']) & (data['Dernier'] > data['MA200']),   # au-dessus des 2
    (data['Dernier'] < data['MA50']) & (data['Dernier'] < data['MA200']),   # en-dessous des 2
]

choix = [1, -1]

data['Trend'] = np.select(conditions, choix, default=0)
```

```python
data.to_csv('./data_prepared.csv', index=False)
```

```python
# Supprimer les lignes où 'Trend' est égal à 0 ou est null
data = data[data['Trend'].notna() & (data['Trend'] != 0)]

# Afficher la taille après nettoyage et les premières lignes
print(f"Nouvelle taille du jeu de données : {len(data)}")
data.head()
```

```python
# 0. Convertir la colonne Date en Datetime et la définir comme index
data['Date'] = pd.to_datetime(data['Date'])
data = data.set_index('Date')

# 1. Filtrer sur la période
mask = (data.index >= '2001-10-11') & (data.index <= '2020-09-02')
data = data.loc[mask].copy()

# 2. Split chronologique 80/10/10
n = len(data)
train_end = int(n * 0.80)
val_end   = int(n * 0.90)

train = data.iloc[:train_end]
val   = data.iloc[train_end:val_end]
test  = data.iloc[val_end:]

print(f"Train : {train.index.min().date()} → {train.index.max().date()}  ({len(train)} lignes)")
print(f"Val   : {val.index.min().date()} → {val.index.max().date()}  ({len(val)} lignes)")
print(f"Test  : {test.index.min().date()} → {test.index.max().date()}  ({len(test)} lignes)")
```
